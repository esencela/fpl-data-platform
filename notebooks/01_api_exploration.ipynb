```python
import requests
import pandas as pd
import json
from pprint import pprint
```

```python
base_url = 'https://fantasy.premierleague.com/api/'
```

```python
def get_json_data(endpoint: str):
    response = requests.get(base_url+endpoint)
    print(response)
    return response.json()
```

```python
data = get_json_data('bootstrap-static')
pprint(data, indent=2)
```

```python
list(data.keys())
```

```python
players = data['elements']
```

```python
df_players = pd.json_normalize(players)
```

```python
ordered_columns = df_players.columns.tolist()
ordered_columns.sort()
ordered_columns
```

```python
columns = ['id', 'element_type', 'first_name', 'second_name', 'web_name', 'total_points', 'goals_scored', 'assists', 'expected_goals', 'expected_assists']
df_players[columns].sort_values('total_points', ascending=False).head(5)
```

```python
teams = data['teams']
df_teams = pd.json_normalize(teams)
```
